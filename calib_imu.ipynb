```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('calibimu.csv')

data =df.to_numpy()
```

```python
def plot_accX(data):
    plt.plot(range(len(data[:,4])), data[:,4])
    plt.legend(['accX'])
    plt.title('plot de la vitesse angulaire en X sur les aquisitions de la calibration de l\'IMU')
    plt.show()

plot_accX(data)
```

```python
moy_vX = np.mean(data[:,4])
print("moyenne de la vitesse angulaire en X : ", moy_vX)

moy_vY = np.mean(data[:,5])
print("moyenne de la vitesse angulaire en Y : ", moy_vY)

moy_vZ = np.mean(data[:,6])
print("moyenne de la vitesse angulaire en Z : ", moy_vZ)

moy_accX = np.mean(data[:,1])
print("moyenne de l'accélération en X : ", moy_accX)
moy_accY = np.mean(data[:,2])
print("moyenne de l'accélération en Y : ", moy_accY)
moy_accZ = np.mean(data[:,3])
print("moyenne de l'accélération en Z : ", moy_accZ)
```

```python
plot_accX(data - moy_vX)
```
